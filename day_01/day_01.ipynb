4. Let's go back a couple of steps. What is a pipeline and what do we use it for?

A pipeline is a chain of tools, the input of one tool is the output of another.
We use pipeline too automate the analysis process, and try to build template across labs for reproducibility, review, and compare.

5. Why do you think nf-core adheres to strict guidelines?

To standardise the input, workflow, and output, that way the data analysis results are comparable to each other. Convention, maintain by different owners.

6. What are the main features of nf-core pipelines?

Documentation, stable releases, open source, CI testing, run anywhere, packaged software.
